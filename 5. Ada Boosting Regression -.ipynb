```python
import pandas as pd
dataset=pd.read_csv("insurance_pre.csv")
```

```python
dataset=pd.get_dummies(dataset,drop_first=True)
dataset=dataset.astype(int)
```

```python
dataset.columns
```

```python
independent=dataset[['age', 'bmi', 'children', 'sex_male','smoker_yes']]
dependent=dataset[['charges']]
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(independent,dependent,test_size=0.22,random_state=42)
```

```python
from sklearn.ensemble import AdaBoostClassifier
regressor=AdaBoostClassifier(n_estimators=100, random_state=0)                        
regressor.fit(X_train,y_train)
y_pred=regressor.predict(X_test)
```

```python
from sklearn.metrics import r2_score
r_score=r2_score(y_test,y_pred)
r_score
```

```python
import pickle
filename="Final_regression_insurance_charg.sav"
pickle.dump(regressor,open(filename,'wb'))
                                
```
