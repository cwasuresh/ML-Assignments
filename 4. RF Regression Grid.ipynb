```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
dataset=pd.read_csv("insurance_pre.csv")
dataset=pd.get_dummies(dataset,drop_first=True)
dataset=dataset.astype(int)
```

```python
dataset.columns
```

```python
independent = dataset.drop('charges', axis=1)
dependent = dataset['charges']
```

```python
from sklearn.model_selection import GridSearchCV
from sklearn.ensemble import RandomForestRegressor
from sklearn.pipeline import Pipeline
pipe = Pipeline([
    ('rf', RandomForestRegressor(random_state=0))
])

param_grid = {
    'rf__n_estimators': [50, 100, 150, 200],
    'rf__criterion': ['squared_error', 'absolute_error', 'friedman_mse', 'poisson'],
    'rf__max_depth': [None, 5, 10, 15, 20],
    'rf__min_samples_split': [2, 5, 10],
    'rf__min_samples_leaf': [1, 2, 4],
    'rf__max_features': [1.0, 'sqrt', 'log2']
}

grid = GridSearchCV(
    estimator=pipe,
    param_grid=param_grid,
    cv=5,
    scoring='r2',
    refit=True,
    verbose=3,
    n_jobs=-1
)

grid.fit(independent, dependent)
```

```python
re = grid.best_score_
print("The R_score for the best parameter {}:".format(grid.best_params_))
```

```python
table = pd.DataFrame(grid.cv_results_)
```
