```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
dataset=pd.read_csv("insurance_pre.csv")
dataset=pd.get_dummies(dataset,drop_first=True)
dataset=dataset.astype(int)
```

```python
dataset.columns
```

```python
independent = dataset.drop('charges', axis=1)
dependent = dataset['charges']
```

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVR

pipe=Pipeline([('scaler', StandardScaler()),('svr', SVR())])
```

```python
from sklearn.model_selection import GridSearchCV
from sklearn.svm import SVR
param_grid={'svr__kernel':['rbf','poly','sigmoid','linear'],
            'svr__C':[10,50,100,250,500,750,1000,1250,1500,2000],
            'svr__gamma':['auto','scale']}
grid=GridSearchCV(estimator=pipe,param_grid=param_grid,cv=5,scoring='r2',refit=True,verbose=3,n_jobs=-1)

grid.fit(independent,dependent)
```

```python
re=grid.cv_results_
print("The R_score for the best parameter {}:".format(grid.best_params_))
```

```python
table =pd.DataFrame.from_dict(re)
```
